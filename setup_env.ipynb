```python
import pandas as pd
import numpy as np
import shutil
```

```python
df = pd.read_csv('data/raw/HAM10000_metadata.csv')
```

```python
CLASS_NAMES = ['akiec', 'bcc', 'bkl', 'df', 'mel', 'nv', 'vasc']
```

```python
#Create Class folder
for split in ['train', 'validation', 'test']:
    for class_name in CLASS_NAMES:
        os.makedirs(f'data/{split}/{class_name}', exist_ok=True)
```

```python
np.random.seed(42)
img_dirs = ['data/raw/HAM10000_images_part_1', 'data/raw/HAM10000_images_part_2']
```

```python
for class_name in CLASS_NAMES:
    class_df = df[df['dx'] == class_name].sample(frac=1, random_state=42)

    total = len(class_df)
    train_size = int(total * 0.7)
    val_size = int(total * 0.15)

    splits = {
        'train': class_df[:train_size],
        'validation': class_df[train_size:train_size + val_size],
        'test': class_df[train_size + val_size:]
    }

    for split_name, split_df in splits.items():
        for _, row in split_df.iterrows():
            img_id = row['image_id']

            for img_dir in img_dirs:
                source = os.path.join(img_dir, f"{img_id}.jpg")
                if os.path.exists(source):
                    dest = f'data/{split_name}/{class_name}/{img_id}.jpg'
                    shutil.copy2(source, dest)
                    break

    print(f"✓ {class_name}")
```
